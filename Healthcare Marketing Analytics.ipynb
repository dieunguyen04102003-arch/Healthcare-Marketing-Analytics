```python
df.ScheduledDay= pd.to_datetime(df.ScheduledDay)
df.AppointmentDay = pd.to_datetime(df.AppointmentDay)

df['Scheduled_Date'] = df.ScheduledDay.dt.date
df['Schedule_Hour'] = df.ScheduledDay.dt.hour
df['Schedule_Weekday'] = df.ScheduledDay.dt.day_name()

df['Appointment_Date'] = df.AppointmentDay.dt.date
df['Appointment_Weekday'] = df.AppointmentDay.dt.day_name()

df["Waiting_days"] = (
    df["AppointmentDay"].dt.normalize()
    - df["ScheduledDay"].dt.normalize()
).dt.days

```

```python
labels = ['Young Adult', 'Adult', 'Middle-Aged', 'Senior']
df['Age_Group'] = pd.qcut(df.Age, q = 4, labels = labels)
```

```python
df.head()
```

```python
def show_up(show):
    if show == 'No':
        return 1
    else:
        return 0
df['Show_up'] = df['No-show'].apply(show_up)
```

```python
df.head()
```

```python
def age_group(age):
    if age < 18:
        return 'Under 18'
    elif age < 30:
        return 'Young Adult'
    elif age < 45:
        return 'Adult'
    elif age < 60:
        return 'Middle-Aged'
    else:
        return 'Senior'

df['Age_Group'] = df['Age'].apply(age_group)
```

```python
df = df.drop(['ScheduledDay', 'AppointmentDay', 'No-show'], axis = 1)
```

```python
def waiting_group(day):
    if day >= 0 and day <= 3:
        return '0-3 days'
    if day >= 4 and day <= 7:
        return '4-7 days'
    if day >= 8 and day <= 14:
        return '8-14 days'
    if day >= 15 and day <= 30:
        return '15-30 days'
    else:
        return '31+ days'
df['Waiting_group'] = df['Waiting_days'].apply(waiting_group)
```

```python
df['Conditions'] = df['Hipertension'] + df['Diabetes'] + df['Alcoholism'] + df['Handcap']
```
